```python
import pandas as pd
orders = pd.read_csv("/content/olist_orders_dataset.csv")
items = pd.read_csv("/content/olist_order_items_dataset.csv")
orders.head()
items.head()
```

###Merge Datasets

```python
df = pd.merge(orders, items, on='order_id')
df.head()
```

###Save Combined Dataset

```python
df.to_csv("ecommerce_data.csv", index=False)
```

###Check Missing Values

```python
df.isnull().sum()
```

###Convert Missing Values in Percentage

```python
df.isnull().sum() / len(df) * 100
```

```python
df["order_approved_at"].fillna(df["order_approved_at"].mode()[0], inplace=True)
df["order_delivered_carrier_date"].fillna(df["order_delivered_carrier_date"].mode()[0], inplace=True)
df["order_delivered_customer_date"].fillna(df["order_delivered_customer_date"].mode()[0], inplace=True)
```

###Monthly Orders Trend

```python
df["order_purchase_timestamp"] = pd.to_datetime(df["order_purchase_timestamp"])
monthly_orders = df.groupby(df["order_purchase_timestamp"].dt.month)\
                                ["order_id"].count()

print(monthly_orders)
```

###Monthly Revenue Trend

```python
monthly_revenue = df.groupby(df["order_purchase_timestamp"].dt.month)\
                                ["price"].sum()

print(monthly_revenue)
```

###Monthly Revenue Trend

```python
import matplotlib.pyplot as plt
monthly_revenue = df.groupby(df["order_purchase_timestamp"].dt.month)\
                    ["price"].sum()

monthly_revenue.plot(kind="line")
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

###Order Status Analysis

```python
status_count = df["order_status"].value_counts()
status_count.plot(kind="bar")

plt.title("Order Status Analysis")
plt.xlabel("Order Status")
plt.ylabel("count")
plt.show()
```

###Top 10 Sellers

```python
top_sellers = df.groupby("seller_id")["price"].sum()\
                 .sort_values(ascending=False).head(10)

top_sellers.plot(kind="bar")
plt.title("Top 10 Sellers")
plt.xlabel("Seller ID")
plt.ylabel("Revenue")
plt.show()
```

###Freight Charges Distribution

```python
df["freight_value"].plot(kind="hist", bins=20)
plt.title("Freight Charges Distribution")
plt.xlabel("Freight Charges")
plt.ylabel("Frequency")
plt.show()
```

###Revenue by Order Status

```python
revenue_status = df.groupby("order_status")["price"].sum()

revenue_status.plot(kind="bar")
plt.title("Revenue by Order Status")
plt.xlabel("Order Status")
plt.ylabel("Revenue")
plt.show()
```

```python
monthly_orders = df.groupby(df["order_purchase_timestamp"].dt.month)\
                      ["order_id"].count()
monthly_orders.plot(kind="line")
plt.title("Monthly Orders Trend")
plt.xlabel("Month")
plt.ylabel("Orders")
plt.show()
```
